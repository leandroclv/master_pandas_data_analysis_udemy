####if, else

```python
numb1 = 30
numb2 = 31
```

```python
if numb1 > numb2:
    print(numb1 + numb2)
else:
    print(numb1 * numb2)
```

```python
if numb1 < numb2 and numb1 > 10:
    print(numb1 + numb2)
else:
    print(numb1 * numb2)
```

```python
age = int(input('How old are you?'))
if age >= 0 and age <= 2:
    print('You are an infant baby')
elif age > 3 and age <= 12:
    print('You are a child')
elif age > 13 and age <= 19:
    print('You are a teen')
elif age > 20 and age <= 35:
    print('You are a young star')
elif age > 36 and age < 50:
    print('You ara a matured man and you shold have wife and kids')
else:
    print('You are a national treasure and we love you')
```

```python
age = int(input('How old are you?'))
if age >= 18 and age < 21:
    print('You can vote')
elif age > 21:
    print('You can vote')
    print('You are matured enough to get married')
else:
    print('You are not matured enough to vote and far from getting married')
```
