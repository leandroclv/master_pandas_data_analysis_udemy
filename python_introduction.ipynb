####Arithmetic operators

- Addition -> x + y
- Subtraction -> x - y
- Multiplication -> x * y
- Division -> x / y
- Modulus -> x % y
- Exponention -> x ** y
- Floor division -> x // y

```python
x = 10
y = 4
```

```python
#Addition
print( x + y)
```

```python
#Subtraction
print(x - y)
```

```python
#Multiplication
print(x * y)
```

```python
#Division
print(x / y)
```

```python
#Modulus
print(x % y)
```

```python
#Exponention
print(x ** y)
```

```python
#Floor division
print(x // y)
```
